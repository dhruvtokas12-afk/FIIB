```python
# Decision making structures
# Nested IF Statement in Pvthon
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
 print("Revenue is high")
 if rating >= 4:
   print("Customer satisfaction is good")
 else:
   print("Customer satisfaction needs improvement")

else:
 print("Revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
  print("Excellent")
elif rating >= 4.0:
    print("Good")
elif rating >= 3.0:
      print("Average")

else:
 print("Needs Improvement")
```

```python
# 1. Correct the casing in number_cols ('order_Value' -> 'Order_Value')
number_cols = ["Items_Count", "Order_Value", "Discount_Pct", "Delivery_Fee", "Distance_Km", "Delivery_Time_Min", "Rating", "Final_Amount"]

# 2. Clean the string columns to numeric values so .describe() works on them
for col in ["Order_Value", "Final_Amount"]:
    if col in df.columns and df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col].astype(str).str.replace(",", "").str.strip(), errors='coerce')

if "Discount_Pct" in df.columns and df["Discount_Pct"].dtype == 'object':
    df["Discount_Pct"] = pd.to_numeric(df["Discount_Pct"].astype(str).str.replace("%", "").str.strip(), errors='coerce')

# 3. Round off and describe
df[number_cols].describe().round(2)
```

```python
# 2.2 Statistics for one important column: Final Amount
amount = df["Final_Amount"]

print("Mean (average): RS", round(amount.mean(), 2))
print("Median (middle): Rs", amount.median())
print("Mode (most common): RS", amount.mode()[0])
print("Minimum: Rs", amount.min())
print("Maximum: RS", amount.max())
print("Standard deviation: RS", round(amount.std(), 2))
print("Range (max - min): RS", amount.max() - amount.min())
```

```python
# Comparison of mean and median
# Calculate average and median values
mean_value = amount.mean()
median_value = amount.median()

print(f"Mean: Rs {mean_value:,.2f}")
print(f"Median: Rs {median_value:,.2f}")

# Compare mean and median
if mean_value > median_value:
    print("A few large orders may be pulling the average up.")
else:
    print("The average is close to or below the median.")
```
